```python
class bank:
    
#Define Constructor
    def __init__(self,acct_no,acct_holder,
                 age,gender,email_id,
                 address,mobile_no,user_id,passcode):
        self.acct_no = acct_no
        self.acct_holder = acct_holder
        self.age = age
        self.gender = gender
        self.email_id = email_id
        self.address = address
        self.mobile_no = mobile_no
        self.user_id = user_id
        self.__passcode = passcode

#Define child class: saving _acct
class saving_acct(bank):
    #Define Constructor
    def __init__(self,interest_rate,
                 available_balance):
        super().__init__(acct_no,acct_holder,
                         age,gender,email_id,address,
                         mobile_no,user_id,passcode)
        self.interest_rate = interest_rate
        self.available_balance = available_balance
        
    def credit(self):
        self.credit = float(input('Enter the credit amount : '))
        print(f'=======================================')
        print(f'account number : {self.acct_no}')
        print(f'account holder : {self.acct_holder}')
        print(f'age : {self.age}')
        print(f'gender : {self.gender}')
        print(f'email id : {self.email_id}')
        print(f'address : {self.address}')
        print(f'mobile number : {self.mobile_no}')
        print(f'credits amount : {self.credit}')
        print(f'interest rate : {self.interest_rate}')
        print(f'available balance : {self.available_balance}')
        print(f'the amount credited : {self.credit}')
        print(f'new updated balance : {self.available_balance + self.credit}')
        print(f'=======================================')
        return self.credit
```
