```python
# Write a program wheather the enter string is palindrome or not

string = input('Enter the string')      # = use to assign the value for string

string1 = string[::-1]      #reverse the given string 

if string == string1:       # == is a condition here
    print('The given is palindrome')
else:
    print('The given string is not a palindrome')   # I have ask for Mom
```
